### Задача 1.8

> В конкурсе по $5$ номинациям участвуют $10$ кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* а) Каждый из вариантов распределения призов представляет собой комбинацию $5$ фильмов из $10$, отличающуюся от других комбинаций как составом фильмов, так и их порядком по номинациям, причем одни и те же фильмы могут повторяться несколько раз.

Следовательно, получаем размещение с повторениями из $10$ элементов по $5$:

$\overline{A}_{10}^{5} = 10^5$

```python
10 ** 5
```

* б) Если по каждой номинации установлены одинаковые призы, то порядок следования фильмов в комбинации $5$ призеров значения не имеет.

Следовательно, получаем сочетание с повторениями из $10$ элементов по $5$:

$\overline{C}_{10}^{5} = C_{10+5-1}^{5} = C_{14}^{5}$

```python
import scipy.special as sc

int(sc.comb(14, 5))
```

### Задача 1.9

> Сколько существует семизначных чисел, состоящих из цифр $4$, $5$ и $6$, в которых цифра $4$ повторяется $3$ раза, а цифры $5$ и $6$ — по $2$ раза?

Каждое семизначное число отличается от другого порядком следования цифр.

При этом цифра $4$ повторяется $3$ раза, цифра $5$ — $2$ раза и цифра $6$ — $2$ раза.

Следовательно, получаем перестановку с повторениями из $7$ элементов:

$P_7(3,2,2)=\frac{7!}{3!\cdot2!\cdot2!}$

```python
import scipy.special as sc

int(sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2)))
```

### Задача 1.10

> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* а) Рассмотрим событие $A$ – получилось слово **ТОР**.

Различные комбинации трех букв из имеющихся шести представляют собой размещения, так как могут отличаться как составом букв, так и порядком их следования.

Общее число случаев:

$n = A_6^3$

Благоприятствует событию $A$ только один случай:

$m = 1$

Следовательно,

$P(A) = \frac{m}{n} = \frac{1}{A_6^3} = \frac{1}{120}$

```python
import scipy.special as sc

round(float(1 / sc.perm(6, 3)), 6)
```

* б) Рассмотрим событие $B$ – получилось слово **ТЕОРИЯ**.

Различные комбинации шести букв из имеющихся шести представляют собой перестановки, так как отличаются только порядком следования букв.

Общее число случаев:

$n = P_6 = 6!$

Благоприятствует событию $B$ только один случай:

$m = 1$

Следовательно,

$P(B) = \frac{m}{n} = \frac{1}{6!} = \frac{1}{720}$

```python
round(float(1 / sc.factorial(6)), 6)
```

### Задача 1.11

> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

Рассмотрим событие $B$ – получилось слово **АНАНАС**.

Общее число случаев представляет собой перестановку из $6$ элементов:

$n = P_6 = 6!$

Число случаев, благоприятствующих событию $B$:

$m = P_3 \cdot P_2 = 3! \cdot 2!$

Следовательно,

$P(B) = \frac{m}{n} = \frac{3! \cdot 2!}{6!} = \frac{1}{60}$

```python
import scipy.special as sc

round(float(sc.factorial(3) * sc.factorial(2) / sc.factorial(6)), 6)
```

### Задача 1.16

> За круглым столом рассаживаются $5$ мужчин и $5$ женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют $5$ супружеских пар.

* а) Рассмотрим событие $A$ – никакие два лица одного пола не сядут рядом.

Общее число способов рассадки $10$ человек:

$n = P_{10} = 10!$

Если женщины займут четные места, то это можно сделать $5!$ способами, а мужчины займут нечетные места также $5!$ способами.

Также возможен обратный вариант: мужчины занимают четные места, а женщины – нечетные.

Поэтому число случаев, благоприятствующих событию $A$:

$m = 2(5!)^2$

Следовательно,

$P(A) = \frac{m}{n} = \frac{2(5!)^2}{10!}$

```python
import scipy.special as sc

round(float(2 * sc.factorial(5) ** 2 / sc.factorial(10)), 6)
```

* б) Рассмотрим событие $B$ – супруги, образующие $5$ супружеских пар, сядут рядом.

Рассматриваем каждую супружескую пару как единый элемент.

Переставить $5$ супружеских пар можно:

$5!$

способами.

При этом внутри каждой из $5$ пар муж и жена могут поменяться местами двумя способами.

По правилу произведения число благоприятствующих случаев:

$m = 5! \cdot 2^5$

Следовательно,

$P(B) = \frac{m}{n} = \frac{5! \cdot 2^5}{10!}$

```python
round(float(sc.factorial(5) * 2 ** 5 / sc.factorial(10)), 6)
```

### Задача 1.17

> В купейный вагон ($9$ купе по $4$ места) семи пассажирам продано $7$ билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Общее число способов выбора $7$ мест из имеющихся $36$:

$n = C_{36}^{7}$

* а) Рассмотрим событие $A$ – пассажиры попали в два купе.

Два купе из $9$ можно выбрать $C_9^2$ способами.

В двух купе находится $8$ мест, из которых нужно выбрать $7$:

$C_8^7$

По правилу произведения число благоприятствующих случаев:

$m_1 = C_9^2 \cdot C_8^7$

Следовательно,

$P(A) = \frac{m_1}{n}
= \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
import scipy.special as sc

p_A = sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7)

print(f"{p_A:.7f}")
```

* б) Рассмотрим событие $B$ – пассажиры попали в семь купе.

Семь купе из $9$ можно выбрать:

$C_9^7$

В каждом из семи купе необходимо выбрать одно место из четырех.

Это можно сделать:

$4^7$

способами.

По правилу произведения число благоприятствующих случаев:

$m_2 = C_9^7 \cdot 4^7$

Следовательно,

$P(B) = \frac{m_2}{n}
= \frac{C_9^7 \cdot 4^7}{C_{36}^7}$

```python
p_B = sc.comb(9, 7) * 4 ** 7 / sc.comb(36, 7)

print(f"{p_B:.5f}")
```

* в) Рассмотрим событие $D$ – пассажиры попали в три купе.

Три купе из $9$ можно выбрать:

$C_9^3$

Семь мест в этих трех купе могут распределиться следующим образом:

$4 + 2 + 1$

$3 + 3 + 1$

$3 + 2 + 2$

С учетом перестановок купе эти варианты можно получить соответственно $6$, $3$ и $3$ способами.

Поэтому число благоприятствующих случаев:

$m_3 =
C_9^3
\left(
6C_4^4 C_4^2 C_4^1
+
3C_4^3 C_4^3 C_4^1
+
3C_4^3 C_4^2 C_4^2
\right)$

Следовательно,

$P(D) = \frac{m_3}{n}$

```python
m_3 = sc.comb(9, 3) * (
    6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)
)

p_D = m_3 / sc.comb(36, 7)

print(f"{p_D:.5f}")
```
