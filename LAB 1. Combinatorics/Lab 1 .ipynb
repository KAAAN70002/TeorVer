### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

* Рассмотрим пункт **а)** – призы различаются по номинациям.

Порядок важен, один и тот же фильм может получить несколько призов. Используем размещения с повторениями:

$N_a = {A}_{10}^{5} = 10^5$

```python
10 ** 5
```

* Рассмотрим пункт **б)** – призы одинаковые.

Порядок не важен, повторения разрешены. Используем сочетания с повторениями:

$N_b = \overline{C}_{10}^{5} = C_{10+5-1}^{5} = C_{14}^{5}$

```python
int(sc.comb(14, 5))
```

### Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?

Используем перестановки с повторениями:

$N = P_7(3,2,2) = \frac{7!}{3! \cdot 2! \cdot 2!}$

```python
int(sc.factorial(7) / (sc.factorial(3) * sc.factorial(2) * sc.factorial(2)))
```

## § 1.6. Непосредственное вычисление вероятностей
---

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим событие $A$ – получилось слово **ТОР**.

Общее число способов выбрать и расположить 3 карточки из 6:

$n = A_6^3 = 6 \cdot 5 \cdot 4$

Благоприятный исход один, поэтому

$P(A) = \frac{1}{A_6^3} = \frac{1}{120}$

```python
round(1 / (6 * 5 * 4), 4)
```

* Рассмотрим событие $B$ – получилось слово **ТЕОРИЯ**.

Все 6 карточек можно переставить $6!$ способами. Благоприятный исход один:

$P(B) = \frac{1}{6!} = \frac{1}{720}$

```python
round(1 / sc.factorial(6), 4)
```

### Задача 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

* Рассмотрим событие $A$ – получилось слово **АНАНАС**.

Общее число перестановок шести карточек равно $6!$.

Перестановка трех карточек с буквой А и двух карточек с буквой Н не меняет слово, поэтому число благоприятных случаев:

$m = 3! \cdot 2!$

Тогда

$P(A) = \frac{3! \cdot 2!}{6!} = \frac{1}{60}$

```python
round(sc.factorial(3) * sc.factorial(2) / sc.factorial(6), 4)
```

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

* Рассмотрим событие $A$ – мужчины и женщины сидят поочередно.

Общее число рассадок:

$n = 10!$

Женщины могут занять четные места $5!$ способами, мужчины — нечетные места $5!$ способами, либо наоборот. Поэтому:

$m = 2(5!)^2$

$P(A) = \frac{2(5!)^2}{10!}$

```python
round(2 * sc.factorial(5) ** 2 / sc.factorial(10), 5)
```

* Рассмотрим событие $B$ – каждая супружеская пара сидит рядом.

Пять пар можно переставить $5!$ способами. Внутри каждой пары муж и жена могут поменяться местами двумя способами:

$m = 5! \cdot 2^5$

$P(B) = \frac{5! \cdot 2^5}{10!}$

```python
round(sc.factorial(5) * 2 ** 5 / sc.factorial(10), 5)
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

Общее число способов выбрать 7 мест из 36:

$n = C_{36}^{7}$

* Рассмотрим событие $A$ – пассажиры попали в **два купе**.

Два купе из девяти выбираются $C_9^2$ способами. Из 8 мест выбранных купе нужно выбрать 7:

$m = C_9^2 \cdot C_8^7$

$P(A) = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
round(sc.comb(9, 2) * sc.comb(8, 7) / sc.comb(36, 7), 7)
```

* Рассмотрим событие $B$ – пассажиры попали в **семь купе**.

Семь купе из девяти выбираются $C_9^7$ способами. В каждом купе выбирается одно из четырех мест:

$m = C_9^7 \cdot 4^7$

$P(B) = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$

```python
round(sc.comb(9, 7) * 4 ** 7 / sc.comb(36, 7), 5)
```

* Рассмотрим событие $D$ – пассажиры попали в **три купе**.

Три купе из девяти выбираются $C_9^3$ способами.

Возможные распределения семи пассажиров по трем купе:

$4+2+1,\quad 3+3+1,\quad 3+2+2$

Поэтому число благоприятных случаев:

$m = C_9^3 \left(6C_4^4C_4^2C_4^1 + 3C_4^3C_4^3C_4^1 + 3C_4^3C_4^2C_4^2\right)$

$P(D) = \frac{m}{C_{36}^7}$

```python
m = sc.comb(9, 3) * (
    6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1)
    + 3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2)
)
round(m / sc.comb(36, 7), 5)
```
